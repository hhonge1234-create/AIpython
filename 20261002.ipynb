```python

def add(x, y):
    return x + y

def sub(x, y):
    return x - y

def mul(x, y):
    return x * y

def div(x, y):
    return x / y


def asmd(x, y, op):
    if op == '+':
        print(f'{x} + {y} = {add(x, y)}')
    elif op == '-':
        print(f'{x} - {y} = {sub(x, y)}')
    elif op == '*':
        print(f'{x} * {y} = {mul(x, y)}')
    elif op == '/':
        print(f'{x} / {y} = {div(x, y):.2f}')


x = int(input('첫 번째 수: '))
y = int(input('두 번째 수: '))
op = input('연산자를 입력하시오: ')

asmd(x, y, op)
```
