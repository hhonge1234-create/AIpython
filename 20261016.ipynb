```python
dan = int(input('단의 정수를 입력하세요: '))
for i in range(1, 10):
  print(f'{dan} x {i} = {dan*i}')
```

```python
n = int(input('정수: '))
count = 0
for a in range(1, n+1):
  if n % a == 0:
    count += 1

print('약수의 개수:', count)
if count == 2:
  print('소수입니다.')
```

```python
sum=0
while True:
  n = int(input("양의 정수를 입력하세요"))
  if n==0:
    break
  sum += n
print(sum)
```

```python
evens = []
for i in range(1, 21):
  if i % 2 == 0:
    evens.append(i)
print(evens)
evens.pop(7)
print(evens)
del evens[5]
print(evens)
print(evens[-1])
print(evens[3])
print(evens[2:5])
```
